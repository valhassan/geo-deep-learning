### Using Git

![git](./imgs/git_and_github.png)

Download and Install 

https://git-scm.com/downloads

- Check if Git is installed
    ```
    git --version
    ```

- Configure your global identity
    ```
    git config --global user.name "Your Name"
    git config --global user.email "you@example.com"
    ```

- Set default branch name
    ```
    git config --global init.defaultBranch main
    ```

- Verify your configuration
    ```
    git config --list
    ```


#### Github

Create an account on Github
https://github.com


#### Using Git and Github

- Generate SSH key
    ```
    ssh-keygen -t ed25519 -C "vic@gmail.com"
    ```

- Start SSH agent
    ```
    eval "$(ssh-agent -s)"
    ```

- Add key
    ```
    ssh-add ~/.ssh/id_ed25519
    ```

- Copy the public key to Github
    ```
    cat ~/.ssh/id_ed25519.pub
    ```

- Test Github Connection
    ```
    ssh -T git@github.com
    ```

<br>

- Coding and Versioning

    1. Clone from Github
        ```
        git clone git@github.com:NRCan/geo-deep-learning.git
        ```

    2. Create a new repo on Github 
        ```
        cd ~/projects
        ```
        ```
        mkdir geo-deep-learning
        ```
        ```
        touch README.md
        ```
        <br>
        
        Intialize git
        ```
        git init
        ```

        Stage the file

        ```
        git add README.md
        ```

        Commit Change
        ```
        git commit -m "Add README"
        ```

        Add Remote and Push
        ```
        git remote add origin git@github.com:USERNAME/geo-deep-learning.git

        git push -u origin main
        ```















### Pytorch


![pytorch](./imgs/pytorch_and_lightning.jpg)

**PyTorch** is a Python library for building and training deep learning models.

1. Tensor Computation

* Works like **NumPy**, handling large arrays of numbers (called *tensors*).
* Can use your **GPU** for massive speed-ups in calculations.

2. Enables deep learning with an autograd system





### Pytorch-Lightning

![lightning](./imgs/lightning_overview.png)

**Pytorch-Lightning** simplifies pytorch enabling reproducible code and scaling
